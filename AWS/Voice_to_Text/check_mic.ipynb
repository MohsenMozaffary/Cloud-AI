# Checks the list of selected inputs

```python


def test_mic(device_index, sr=48000, seconds=2):
    print("Testing device:", device_index, sd.query_devices(device_index)["name"])
    audio = sd.rec(int(seconds * sr), samplerate=sr, channels=1, dtype="float32", device=device_index)
    sd.wait()
    rms = float(np.sqrt(np.mean(audio**2)))
    mx = float(np.max(np.abs(audio)))
    print(f"RMS={rms:.6f}  MAX={mx:.6f}")

# (talk while it records)
for dev in [17, 8, 1, 18, 2, 27]:
    try:
        test_mic(dev, sr=48000)
    except Exception as e:
        print("Device", dev, "failed:", e)
```
