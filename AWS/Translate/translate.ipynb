```python
import boto3
from botocore.exceptions import BotoCoreError, ClientError

def aws_translate_text(
    text: str,
    source_lang: str,
    target_lang: str,
    region_name: str = "us-east-1",
) -> str:
    """
    Translate text using Amazon Translate.

    Args:
        text: Input text (UTF-8). Max 10,000 bytes per request.
        source_lang: Source language code.
        target_lang: Target language code.
        region_name: AWS Translate region.

    Returns:
        Translated text.
    """
    if not text or not text.strip():
        return ""

    if len(text.encode("utf-8")) > 10_000:
        raise ValueError("Text is too long (max 10,000 UTF-8 bytes for TranslateText).")

    session = boto3.Session(region_name=region_name)

    client = session.client("translate")

    try:
        resp = client.translate_text(
            Text=text,
            SourceLanguageCode=source_lang,
            TargetLanguageCode=target_lang,
        )
        return resp["TranslatedText"]
    except (BotoCoreError, ClientError) as e:
        raise RuntimeError(f"Amazon Translate failed: {e}") from e
```

```python
print(aws_translate_text("سلام، حالت چطوره؟", "auto", "en"))
print(aws_translate_text("How are you?", "auto", "fa"))
```
